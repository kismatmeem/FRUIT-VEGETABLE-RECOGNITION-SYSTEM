```python
##starting visualization then  prediction
import cv2
image_path= "/content/drive/MyDrive/fruit-veg-recognition/test/banana/Image_1.jpg"
img=cv2.imread(image_path)
plt.imshow(img)
plt.title("test image")
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
image= tf.keras.preprocessing.image.load_img(image_path,target_size=(64,64))
input_arr=tf.keras.preprocessing.image.img_to_array(image)
input_arr =np.array([input_arr])               ##single image to batch
predictions=cnn.predict(input_arr)
```

```python
print(predictions)
print(max(predictions[0]))
```
